```python
#Write a program to check whether a number is positive, negative, or zero.

number = int(input("Enter the number:"))

if(number>0) :
    print("number is positive")
elif(number<0) :
    print("number is negative")
else :
    print("number is zero")
```

```python
# Write a program to check whether a student passed or failed.
student_name= str(input('Enter the name of student '))
student_marks= int(input('Enter the markes '))
if(student_marks>=35) :
    print(student_name,'student passed')
else :
    print(student_name,'student failed')
```

```python
#Write a program to print numbers from 1 to 100 using a for loop.
for i in range(1,101):
  print(i, end=" ")
```

```python
#Write a program to print numbers from 100 to 1 using a while loop.
number=100
while(number>=1) :
    print(number, end=" ")
    number-=1
```

```python
# Write a program to calculate the sum of numbers from 1 to 100.
result=0
for i in range(1,101) :
  result+=i

print(result)
```

```python
print("Select option :")
print("1. Sum")
print("2. Substraction")
print("3. Divide")
print("4. Multiplication")

choise= int(input("Enter your choise: "))
first_num= int(input("Enter first number: "))
sec_num= int(input("Enter second number: "))
match choise:
  case 1:
    print('Result: ',first_num+sec_num)
  case 2:
    print("Result: ", first_num-sec_num)
  case 3:
    print("Result: ", first_num/sec_num)
  case 4:
    print("Result:", first_num*sec_num)
  case _:
    print("Enter valid option")
```

```python
#Write a program to print all even numbers from 1 to 50.

for i in range(1,51):
  if(i%2==0):
    print(i,end=" ")
```

```python
#Write a program to skip numbers divisible by 3 using continue.

for i in range(1,51):
  if(i%3==0):
    continue;

  print(i,end=" ")
```

```python
#Write a program to stop searching when a particular product is found using break.

products=['Mouse', 'Keyboard', 'Charger','Laptop' ]
product_name=str(input('Enter product name: ').strip())
for product in products:
  if(product.lower() == product_name.lower()):
    print('Product found:',product)
    break;
else:
    print('Search diff product')


```
