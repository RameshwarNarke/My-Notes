```python
#1. Decreasing Star Triangle
for i in range(1,6):
  for j in range(i,6):
    print('*', end="")
  print()
```

```python
#2 Increasing Star Triangle
for i in range(1,6):
  for j in range(1,i+1):
    print('*',end="")
  print()
```

```python
#Number Triangle

for i in range(1,6):
  for j in range(1,i+1):
    print(j,end="")

  print()
```
