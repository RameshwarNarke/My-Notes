```python
employee_name=input("Enter Employee Name: ")
basic_salary=int(input("Enter Basic Salary: "))


def calculate_basic_salary(basic_salary, allowance=5000):
  return basic_salary+allowance

def calculate_total_salary(*ammounts):
  total_salary=0
  for amount in ammounts:
      total_salary+= amount
  return total_salary

def print_employee_details(**kwargs):
  print("Employee Name: ", kwargs["employee_name"])
  print("Basic Salary: ", kwargs["basic_salary"])
  print("Allowances: ", kwargs["allowance"])
  print("Total Salary: ",kwargs["total_salary"])



salary_with_allowance=calculate_basic_salary(basic_salary)

total_salary= calculate_total_salary(salary_with_allowance,2000)

print_employee_details(employee_name=employee_name,basic_salary=basic_salary,allowance=5000,total_salary=total_salary)


```
